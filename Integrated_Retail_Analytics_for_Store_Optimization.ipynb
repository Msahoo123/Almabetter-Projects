# **1. Anomaly Detection in Sales Data:**
Anomaly detection in sales data is a crucial technique in retail analytics, helping businesses identify unusual patterns that may indicate operational issues, fraud, or unexpected customer behavior. By spotting anomalies early, retailers can optimize inventory, improve forecasting, and enhance decision‑making.

**🔎 Why Anomaly Detection Matters**

 * Customer behavior: Detect shifts in buying trends early.

 * Inventory optimization: Spot irregular demand to adjust stock levels.

 * Fraud detection: Identify suspicious transactions.

 * Operational efficiency: Prevent bottlenecks caused by unexpected anomalies.

**🐍 Python Code Example**

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Sample daily sales data
data = {
    "day": pd.date_range(start="2026-08-01", periods=30, freq="D"),
    "sales": [200, 220, 210, 205, 215, 500, 225, 230, 240, 235,
              250, 260, 255, 245, 600, 270, 280, 275, 265, 290,
              295, 300, 310, 305, 1000, 320, 330, 325, 315, 340]
}
df = pd.DataFrame(data)

# Apply Isolation Forest
model = IsolationForest(contamination=0.1, random_state=42)
df["anomaly"] = model.fit_predict(df[["sales"]])

# Separate normal vs anomalies
normal = df[df["anomaly"] == 1]
anomalies = df[df["anomaly"] == -1]

# Plot
plt.figure(figsize=(10,6))
plt.plot(df["day"], df["sales"], label="Sales", color="blue")
plt.scatter(normal["day"], normal["sales"], color="green", label="Normal")
plt.scatter(anomalies["day"], anomalies["sales"], color="red", label="Anomaly", marker="x", s=100)
plt.title("Anomaly Detection in Sales Data")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# **2. Identify unusual sales patterns across stores and departments.**

 In retail analytics, detecting anomalies across stores and departments is essential for optimizing operations. Unusual sales patterns may indicate:

 * Inventory misalignment (stockouts or overstocking)

 * Fraudulent activity (suspicious spikes or drops)

 * Seasonal demand shifts (holiday surges, festival effects)

 * Operational inefficiencies (poor staffing, mispricing)

**🐍 Python Code Example**

Here’s a simple workflow using Isolation Forest to flag anomalies in sales across multiple stores and departments:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest

# Sample dataset: sales across stores and departments
data = {
    "store": ["A","A","A","B","B","B","C","C","C","C"],
    "department": ["Electronics","Clothing","Groceries",
                   "Electronics","Clothing","Groceries",
                   "Electronics","Clothing","Groceries","Furniture"],
    "sales": [2000, 1500, 3000, 2200, 8000, 3100, 2100, 1600, 3050, 15000]
}

df = pd.DataFrame(data)

# Apply Isolation Forest for anomaly detection
model = IsolationForest(contamination=0.2, random_state=42)
df["anomaly"] = model.fit_predict(df[["sales"]])

# Separate normal vs anomalies
normal = df[df["anomaly"] == 1]
anomalies = df[df["anomaly"] == -1]

# Heatmap of sales
plt.figure(figsize=(8,6))
pivot = df.pivot("store","department","sales")
sns.heatmap(pivot, annot=True, cmap="Blues")
plt.title("Sales Heatmap Across Stores and Departments")
plt.show()

# Scatter plot highlighting anomalies
plt.figure(figsize=(8,6))
plt.scatter(normal["department"], normal["sales"], color="green", label="Normal")
plt.scatter(anomalies["department"], anomalies["sales"], color="red", label="Anomaly", marker="x", s=100)
plt.title("Anomaly Detection in Department Sales")
plt.xlabel("Department")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# **4. Implement anomaly handling strategies to clean the data for further analysis.**

 When unusual sales patterns are detected across stores and departments, the next step is to clean the data so that further analysis (forecasting, optimization, reporting) is reliable.

**🔎 Common Strategies**

  * Remove anomalies → Drop extreme outliers that distort trends.

  * Cap values → Replace anomalies with upper/lower thresholds.

  * Impute values → Replace anomalies with averages, medians, or model predictions.

  * Flag anomalies → Keep anomalies but mark them for special analysis.

  * Smooth data → Apply rolling averages to reduce noise.

**🐍 Python Code Example**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample sales data with anomalies
data = {
    "date": pd.date_range(start="2026-08-01", periods=15, freq="D"),
    "sales": [200, 220, 210, 205, 215, 500, 225, 230, 240, 235,
              250, 260, 1000, 270, 280]
}
df = pd.DataFrame(data)

# Step 1: Detect anomalies (simple threshold for demo)
threshold = 400
df["anomaly"] = df["sales"] > threshold

# Step 2: Handle anomalies (replace with median)
median_sales = df["sales"].median()
df["clean_sales"] = np.where(df["anomaly"], median_sales, df["sales"])

# Step 3: Plot before vs after cleaning
plt.figure(figsize=(10,6))
plt.plot(df["date"], df["sales"], label="Original Sales", color="blue")
plt.scatter(df[df["anomaly"]]["date"], df[df["anomaly"]]["sales"],
            color="red", label="Anomaly", marker="x", s=100)
plt.plot(df["date"], df["clean_sales"], label="Cleaned Sales", color="green")
plt.title("Anomaly Handling in Sales Data")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# **5. Time-Based Anomaly Detection:**

Time‑based anomaly detection focuses on identifying unusual sales patterns over time series. Instead of looking only at raw values, it considers temporal trends such as daily, weekly, or seasonal cycles. This is especially useful in retail, where sales naturally fluctuate due to holidays, promotions, or economic shifts.

**🔎 Why It Matters**

 * Seasonality detection → Spot deviations from expected holiday or weekend patterns.

 * Trend monitoring → Identify sudden drops or spikes against long‑term growth.

 * Operational alerts → Trigger warnings when sales deviate from forecasts.

 * Forecast validation → Ensure predictive models remain accurate.

**🐍 Python Code Example**

Here’s a simple workflow using rolling averages to detect anomalies in time‑series sales data:

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample daily sales data
data = {
    "date": pd.date_range(start="2026-08-01", periods=30, freq="D"),
    "sales": [200, 220, 210, 205, 215, 500, 225, 230, 240, 235,
              250, 260, 255, 245, 600, 270, 280, 275, 265, 290,
              295, 300, 310, 305, 1000, 320, 330, 325, 315, 340]
}
df = pd.DataFrame(data)

# Rolling mean and standard deviation
df["rolling_mean"] = df["sales"].rolling(window=5).mean()
df["rolling_std"] = df["sales"].rolling(window=5).std()

# Flag anomalies (outside mean ± 2*std)
df["anomaly"] = (df["sales"] > df["rolling_mean"] + 2*df["rolling_std"]) | \
                (df["sales"] < df["rolling_mean"] - 2*df["rolling_std"])

# Plot
plt.figure(figsize=(10,6))
plt.plot(df["date"], df["sales"], label="Sales", color="blue")
plt.plot(df["date"], df["rolling_mean"], label="Rolling Mean", color="orange")
plt.scatter(df[df["anomaly"]]["date"], df[df["anomaly"]]["sales"],
            color="red", label="Anomaly", marker="x", s=100)
plt.title("Time-Based Anomaly Detection in Sales Data")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

# **8. Use time-series analysis for understanding store and department performance over time.**

Time‑series analysis is a powerful tool in retail analytics, allowing managers to track performance trends across stores and departments over time. By modeling sales data as a sequence, retailers can uncover seasonal patterns, detect anomalies, and forecast future demand.

# **9. Data Preprocessing and Feature Engineering:**

In any data science project, especially for retail analytics, the quality of insights depends heavily on how well the data is prepared. Raw sales data often contains missing values, duplicates, anomalies, and categorical variables that need transformation. Preprocessing ensures clean, consistent input, while feature engineering creates new variables that improve model performance.

**🔎 Key Steps in Data Preprocessing**

Handle missing values → Imputation (mean, median, mode) or removal.

Remove duplicates → Ensure unique transactions.

Normalize/scale → Standardize numerical features for algorithms.

Encode categorical variables → One‑hot or label encoding for departments, stores, etc.

Detect anomalies → Identify outliers that distort analysis.

**🛠️ Feature Engineering Examples**

Time-based features → Day of week, month, holiday flag.

Lag features → Previous day/week sales for forecasting.

Rolling averages → Smooth trends and capture seasonality.

Interaction features → Store × Department combinations.

Derived metrics → Sales per customer, discount percentage.

**🐍 Python Code Example**

```python
import pandas as pd
import numpy as np

# Sample sales dataset
data = {
    "date": pd.date_range(start="2026-08-01", periods=10, freq="D"),
    "store": ["A","A","B","B","C","C","A","B","C","A"],
    "department": ["Electronics","Clothing","Groceries","Clothing",
                   "Furniture","Groceries","Electronics","Clothing",
                   "Furniture","Groceries"],
    "sales": [200, np.nan, 300, 250, 1000, 280, 220, 260, 1500, 270]
}
df = pd.DataFrame(data)

# Step 1: Handle missing values (impute with median)
df["sales"].fillna(df["sales"].median(), inplace=True)

# Step 2: Encode categorical variables
df = pd.get_dummies(df, columns=["store","department"], drop_first=True)

# Step 3: Feature engineering
df["day_of_week"] = df["date"].dt.day_name()
df["is_weekend"] = df["date"].dt.dayofweek >= 5
df["lag_sales"] = df["sales"].shift(1)
df["rolling_avg"] = df["sales"].rolling(window=3).mean()

print(df.head())
```

# **10. Handle missing values, especially in the MarkDown data.**

In retail datasets (like the Rossmann sales dataset), the MarkDown fields often contain missing values. These represent promotional markdowns applied to products, and missing entries can distort analysis if not treated properly.

**🔎 Strategies for Handling Missing MarkDown Values**

Imputation with zero → If missing means “no markdown applied,” replace with 0.

Median/mean imputation → Fill gaps with typical markdown values.

Forward/backward fill → Use nearby values to approximate missing entries.

Domain-specific rules → For example, treat missing as “no promotion” unless flagged otherwise.

Feature engineering → Create a binary flag (is_markdown_missing) to capture the effect of missingness itself.

**🐍 Python Code Example**

```python
import pandas as pd
import numpy as np

# Sample dataset with missing MarkDown values
data = {
    "date": pd.date_range(start="2026-08-01", periods=10, freq="D"),
    "sales": [2000, 2200, 2100, 2500, 2700, 2800, 2900, 3000, 3100, 3200],
    "MarkDown1": [np.nan, 100, np.nan, 200, np.nan, 150, np.nan, 120, np.nan, 180]
}
df = pd.DataFrame(data)

# Strategy 1: Replace missing with 0 (no markdown)
df["MarkDown1_zero"] = df["MarkDown1"].fillna(0)

# Strategy 2: Median imputation
median_val = df["MarkDown1"].median()
df["MarkDown1_median"] = df["MarkDown1"].fillna(median_val)

# Strategy 3: Add missing flag
df["MarkDown1_missing_flag"] = df["MarkDown1"].isna().astype(int)

print(df)
```

# **15. Market Basket Analysis:**

Market Basket Analysis (MBA) is a technique used in retail analytics to uncover associations between products that customers frequently purchase together. It helps retailers design cross-selling strategies, optimize store layouts, and improve promotion planning.

**🔎 Key Concepts**

Association rules → Identify product relationships (e.g., “If milk, then bread”).

Support → Frequency of itemset occurrence.

Confidence → Likelihood of buying item B when item A is bought.

Lift → Strength of association compared to random chance.

**🐍 Python Code Example**

Here’s a simple workflow using the Apriori algorithm for Market Basket Analysis:

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Sample transaction dataset
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Eggs'],
    ['Milk', 'Bread', 'Eggs', 'Butter'],
    ['Milk', 'Bread'],
    ['Eggs', 'Butter']
]

# Convert to one-hot encoded DataFrame
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

# Apply Apriori algorithm
frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

print(rules[['antecedents','consequents','support','confidence','lift']])
```

```python
import pandas as pd

# Sample correlation matrix (from department sales analysis)
corr_data = {
    "Beverages": {"Snacks": 0.85, "Frozen": 0.70, "Household": 0.40},
    "Snacks": {"Beverages": 0.85, "Frozen": 0.65, "Household": 0.35},
    "Frozen": {"Beverages": 0.70, "Snacks": 0.65, "Household": 0.50},
    "Household": {"Beverages": 0.40, "Snacks": 0.35, "Frozen": 0.50}
}

df = pd.DataFrame(corr_data)

# Define threshold for strong associations
threshold = 0.6

# Extract cross-selling pairs
cross_sell_pairs = []
for dept1 in df.columns:
    for dept2 in df.index:
        if dept1 != dept2 and df.loc[dept2, dept1] >= threshold:
            cross_sell_pairs.append((dept1, dept2, df.loc[dept2, dept1]))

print("Recommended Cross-Selling Pairs:")
for pair in cross_sell_pairs:
    print(f"{pair[0]} ↔ {pair[1]} (Correlation: {pair[2]:.2f})")
```

Feature-rich models → Incorporate markdowns, holidays, regional demand, and store type.

**🐍 Python Code Example**

Here’s a workflow using Prophet to forecast weekly sales for each store and department:

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt

# Sample weekly sales dataset
data = {
    "date": pd.date_range(start="2025-01-01", periods=30, freq="W"),
    "store": ["A"]*15 + ["B"]*15,
    "department": ["Electronics"]*10 + ["Clothing"]*5 + ["Groceries"]*10 + ["Furniture"]*5,
    "sales": [2000,2200,2100,2500,2700,2800,2900,3000,3100,3200,
              1500,1600,1700,1800,1900,8000,8200,8100,8300,8500,
              3050,3100,3150,3200,3250,15000,15200,15100,15300,15400]
}
df = pd.DataFrame(data)

# Forecast weekly sales for each store-department
forecasts = {}
for (store, dept), group in df.groupby(["store","department"]):
    ts = group[["date","sales"]].rename(columns={"date":"ds","sales":"y"})
    model = Prophet()
    model.fit(ts)
    future = model.make_future_dataframe(periods=8, freq="W")
    forecast = model.predict(future)
    forecasts[(store,dept)] = forecast

    # Plot forecast
    model.plot(forecast)
    plt.title(f"Weekly Sales Forecast: Store {store}, Dept {dept}")
    plt.show()
```

# **28. Segmentation Quality Evaluation:**

Evaluating the quality of store and department segmentation ensures that clusters are meaningful, actionable, and aligned with business goals. A good segmentation should not only separate entities mathematically but also provide strategic insights for inventory, marketing, and demand forecasting.

**🔎 Key Evaluation Metrics**

Silhouette score → Measures how well each store/department fits within its cluster compared to others.

Davies–Bouldin index → Evaluates cluster separation and compactness (lower is better).

Calinski–Harabasz index → Assesses variance ratio between clusters (higher is better).

Business interpretability → Do clusters make sense in terms of markdown intensity, regional demand, and sales patterns?

Actionability → Can strategies (inventory, promotions, personalization) be tailored to each segment?

**🐍 Python Example: Evaluating Segmentation Quality**

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

# Sample dataset
data = {
    "sales": [2000, 8000, 2100, 15000, 3000, 500],
    "markdown": [100, 500, 120, 1000, 200, 50],
    "region_Urban": [1,0,1,0,1,0]
}
df = pd.DataFrame(data)

# Apply clustering
kmeans = KMeans(n_clusters=3, random_state=42)
labels = kmeans.fit_predict(df)

# Evaluate segmentation quality
sil_score = silhouette_score(df, labels)
db_index = davies_bouldin_score(df, labels)
ch_index = calinski_harabasz_score(df, labels)

print(f"Silhouette Score: {sil_score:.2f}")
print(f"Davies-Bouldin Index: {db_index:.2f}")
print(f"Calinski-Harabasz Index: {ch_index:.2f}")
```
